```python
import numpy as np
```

```python
def init_weights(n: int, m: int, k: int):
    W_xh = np.random.randn(m, n) * np.sqrt(2 / (m + n)) # variance should be 2 / (n_in + n_out)
    W_hh = np.random.randn(m, m) * np.sqrt(2 / (m + m))
    W_hy = np.random.randn(k, m) * np.sqrt(2 / (k + m))
    b_h = np.zeros((m, 1))
    b_y = np.zeros((k, 1))
    return W_xh, W_hh, W_hy, b_h, b_y

class Sigmoid:
    @staticmethod
    def f(x):
        return 1 / (1 + np.exp(-x))
    
    def derivative(x):
        return Sigmoid.f(x) * (1 - Sigmoid.f(x))
```

```python
class RNN:
    def __init__(self, n: int, m: int, k: int):
        self.n = n
        self.m = m
        self.k = k
        self.W_xh, self.W_hh, self.W_hy, self.b_h, self.b_y = init_weights(n, m, k)

    def forward_step(self, x_t, h_tminus1, activation_func=Sigmoid.f):
        assert x_t.shape == (self.n, 1)
        a_t = self.W_xh @ x_t + self.W_hh @ h_tminus1 + self.b_h
        h_t = activation_func(a_t)
        assert h_t.shape == (self.m, 1)
        y_pred_t = self.W_hy @ h_t + self.b_y
        assert y_pred_t.shape == (self.k, 1)

        return a_t, h_t, y_pred_t

    def forward(self, X, h_0):
        # Shape of X is (T, n) where each row is one time step
        T  X.shape[0]
        x_list = []
        a_list = []
        h_list = [h_0]
        y_pred_list = []

        for t in range(T):
            x_t = X[t].reshape(-1, 1)
            assert x_t.shape[0] == self.n
            h_tminus1 = h_list[-1]
            x_list.append(x_t)

            a_t, h_t, y_pred_t = self.forward_step(x_t, h_tminus1)
            a_list.append(a_t)
            h_list.append(h_t)
            y_pred_list.append(y_pred_t)
        
        return x_list, a_list, h_list, y_pred_list
```
